# What is the analysis of Country?

```python
country = df['Your Current Country.'].value_counts()
label = country.index
count = country.values
fig = go.Figure(data = [go.Pie(labels = label, values=count)])
fig.update_layout(title_text='Current Country', autosize=False, width=700, height=700)
fig.update_traces(hoverinfo='label+value', textinfo='percent', textfont_size=25, marker=dict(colors=['red', 'lightgreen', 'pink', 'brown'], line=dict(color='black', width=1)))
fig.show()
```

# What are the factors influencing the career aspirations of GenZ

```python
questions = df['Which of the below factors influence the most about your career aspirations ?'].value_counts()
label = questions.index
count = questions.values
fig = go.Figure(data = [go.Pie(labels = label, values=count)])
fig.update_layout(title_text='Factors Influencing Career Aspirations', autosize=False, width=800, height=700)
fig.update_traces(hoverinfo='label+value', textinfo='percent', textfont_size=25, marker=dict(colors=['gold', 'lightgreen', 'pink', 'brown'], line=dict(color='black', width=2)))
fig.show()
```

# How many want to pursue higher education outside india with their investment

```python
outside_education = df['Would you definitely pursue a Higher Education / Post Graduation outside of India ? If only you have to self sponsor it.'].value_counts()
label = outside_education.index
count = outside_education.values
fig = go.Figure(data = [go.Pie(labels=label, values=count)])
fig.update_layout(title='Number of Students Want to Pursue Higher Education Outside Inida With Their Investment', autosize=False, height=600, width=800)
fig.update_traces(hoverinfo='label+value', textinfo='percent', textfont_size=25, marker=dict(colors=['gold', 'lightgreen', 'pink'], line=dict(color='black', width=2)))
fig.show()
```

# How Likely GenZ is to work for one company for three years or more

```python
people = df['How likely is that you will work for one employer for 3 years or more ?'].value_counts()
label = people.index
count = people.values
fig = go.Figure(data = [go.Pie(labels=label, values=count)])
fig.update_layout(title='Employees that will work for one employer for 3 years or more', autosize=False, width=800, height=600)
fig.update_traces(hoverinfo='label+value', textinfo='percent', textfont_size=20, marker=dict(colors=['#0564fc', '#5cf72d', '#f72d7e'], line=dict(color='black', width=2)))
fig.show()
```

# What is the preferred working Environment of GenZ?

```python
environment = df['What is the most preferred working environment for you.'].value_counts()
label = environment.index
count = environment.values
fig = go.Figure(data = [go.Pie(labels=label, values=count)])
fig.update_layout(title='Most Preferred Working Environment of GenZ', autosize=False, width=900, height=600)
fig.update_traces(hoverinfo='label+value', textinfo='percent', textfont_size=20, marker=dict(colors=['rgb(56, 75, 126)', 'rgb(18, 36, 37)', 'rgb(34, 53, 101)', 'rgb(36, 55, 57)', 'rgb(6, 4, 4)'] ,line=dict(color='black', width=2)))
fig.show()
```

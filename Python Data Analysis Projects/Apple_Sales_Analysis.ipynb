# Top 10 Iphones Which are sold most

```python
highest_rated = df.sort_values(by=['Star Rating'], ascending=False)
highest_rated_products = highest_rated.head(10)
print(highest_rated_products['Product Name'])
```

# Lets have a look on the highest rated phone on Flipkart

```python
iphones = highest_rated_products['Product Name'].value_counts()
labels = iphones.index
counts = highest_rated_products['Number Of Ratings']
fig = px.bar(highest_rated_products, x=labels, y=counts, title='Number of Ratings of Highest Rated Iphones')
fig.show()
```

```python
iphones = highest_rated_products['Product Name'].value_counts()
labels = iphones.index
counts = highest_rated_products['Number Of Reviews']
fig = px.bar(highest_rated_products, x=labels, y=counts, title='Number of Reviews of Highest Rated Iphones')
fig.show()
```

```python
product_list = []
for name in df['Product Name']:
    p = name.split(' (')
    product_list.append(p[0])

df['Product'] = product_list
```

```python
sns.barplot(data=df, x='Product', y='Number Of Ratings')
plt.xticks(rotation='vertical')
plt.title('Number of Ratings of People According to Products')
# plt.grid()
plt.show()
```

```python
sns.barplot(data=df, x='Product', y='Number Of Reviews')
plt.xticks(rotation='vertical')
plt.title('Number of Reviews of People According to Products')
# plt.grid()
plt.show()
```

```python
sns.barplot(data=df, x='Product', y='Star Rating')
plt.xticks(rotation='vertical')
plt.title('Star Rating According to Products')
# plt.grid()
plt.show()
```
